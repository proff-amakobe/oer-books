## Step 1 — Python Version

```python
import sys

MIN_PYTHON = (3, 9)

py_version = sys.version_info
print(f"Detected Python version: {py_version.major}.{py_version.minor}.{py_version.micro}")

if (py_version.major, py_version.minor) >= MIN_PYTHON:
    print(f"\u2713 Python version supported (>= {MIN_PYTHON[0]}.{MIN_PYTHON[1]})")
else:
    print(f"\u2717 Python {MIN_PYTHON[0]}.{MIN_PYTHON[1]}+ is required. Please upgrade your Python installation.")
```

## Step 2 — Core Scientific Packages

These packages are used from Chapter 1 onward for data handling, visualization, and classical machine learning baselines.

```python
import importlib

core_packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "matplotlib": "matplotlib",
    "sklearn": "scikit-learn",
    "PIL": "Pillow",
}

results = {}

for import_name, pip_name in core_packages.items():
    try:
        module = importlib.import_module(import_name)
        version = getattr(module, "__version__", "unknown version")
        print(f"\u2713 {pip_name} installed ({version})")
        results[pip_name] = True
    except ImportError:
        print(f"\u2717 {pip_name} is NOT installed.")
        print(f"  Install it with: pip install {pip_name}")
        results[pip_name] = False
```

## Step 3 — Deep Learning Frameworks

This book primarily uses **TensorFlow/Keras** in setup and early code examples, and references **PyTorch** as an accepted alternative in several later hands-on activities (the choice is noted per activity). You do not need both. Install at least one before Chapter 3.

```python
import importlib

dl_frameworks = {"tensorflow": "tensorflow", "torch": "torch"}
dl_installed = {}

for import_name, pip_name in dl_frameworks.items():
    try:
        module = importlib.import_module(import_name)
        version = getattr(module, "__version__", "unknown version")
        print(f"\u2713 {pip_name} installed ({version})")
        dl_installed[pip_name] = True
    except ImportError:
        print(f"\u2717 {pip_name} is not installed.")
        dl_installed[pip_name] = False

if not any(dl_installed.values()):
    print()
    print("No deep learning framework detected. This is fine for Chapters 1-2 (which use only")
    print("scikit-learn and visual/interactive tools), but install one before Chapter 3:")
    print("  pip install tensorflow      # recommended default for this book")
    print("  pip install torch torchvision   # alternative, used in several hands-on notebooks")
```

## Step 4 — GPU / CUDA Availability

A GPU speeds up training but is **not required** for the great majority of this book's exercises, which are deliberately sized to run on a CPU in a few minutes. This check simply reports what is available so you know what to expect.

```python
gpu_available = False
gpu_details = "No deep learning framework installed to check GPU support."

try:
    import tensorflow as tf
    gpus = tf.config.list_physical_devices('GPU')
    if gpus:
        gpu_available = True
        gpu_details = f"TensorFlow sees {len(gpus)} GPU(s): {[g.name for g in gpus]}"
    else:
        gpu_details = "TensorFlow is installed but no GPU was detected."
except ImportError:
    try:
        import torch
        if torch.cuda.is_available():
            gpu_available = True
            gpu_details = f"PyTorch sees {torch.cuda.device_count()} GPU(s): {torch.cuda.get_device_name(0)}"
        else:
            gpu_details = "PyTorch is installed but CUDA is not available."
    except ImportError:
        pass

status = "\u2713" if gpu_available else "\u2139"
print(f"{status} CUDA available: {'Yes' if gpu_available else 'No'}")
print(f"  {gpu_details}")
if not gpu_available:
    print("  This is okay. Most introductory notebooks in this book run comfortably on CPU.")
    print("  Chapters that benefit meaningfully from a GPU (e.g. Chapter 12, Stable Diffusion)")
    print("  say so explicitly and provide guidance for using a free Colab GPU runtime.")
```

## Step 5 — End-to-End Pipeline Check

This is the real test: generate a tiny synthetic dataset, fit a small model, and produce a prediction. If this cell runs without error, your environment can support the hands-on activities in this book, independent of which deep learning framework you chose.

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split

SEED = 42
np.random.seed(SEED)

# Generate a small synthetic dataset
X, y = make_classification(
    n_samples=200, n_features=8, n_informative=5,
    n_classes=2, random_state=SEED
)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=SEED
)

# Fit a tiny model
clf = MLPClassifier(hidden_layer_sizes=(16,), max_iter=500, random_state=SEED)
clf.fit(X_train, y_train)

accuracy = clf.score(X_test, y_test)
print(f"Pipeline complete. Tiny model test accuracy: {accuracy:.2f}")
print("\u2713 Full inference pipeline works: data generation \u2192 training \u2192 prediction \u2192 evaluation.")
```
